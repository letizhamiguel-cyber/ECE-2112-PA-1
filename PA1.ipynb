# A. WORD ROTATION PROBLEM 

```python
def rotate_word(text):
    return text[1:] + text[0]
```

```python
print(rotate_word ("birthday"))
print(rotate_word ("heart"))
print(rotate_word ("know"))
print(rotate_word ("J"))
```

# B. USERNAME BUILDER PROBLEM 

```python
def make_username (first_name, last_name):
    
    build_first = first_name.lower().replace(" ", "")
    build_last = last_name.lower().replace(" ", "")
    
    return build_first + "." + build_last
```

```python
print(make_username("Electronics", "Engineering"))
print(make_username("Mysterious", "Companionship"))
print(make_username("Farmer Oak", "Thomas Hardy"))
```

# C. BOOKEND SWAP PROBLEM

```python
def swap_bookends (items):
    first, *middle, last = items
    
    return [last] + middle + [first]
```

```python
print(swap_bookends ([4, 5, 6, 7, 8, 9]))
print(swap_bookends(["math", "piano", "book"]))
print(swap_bookends([0, 1]))
```
